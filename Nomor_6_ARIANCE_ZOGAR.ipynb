```python
# NOMOR 6 - ARIANCE ZOGAR - Kelas A

import numpy as np

# 2x + y + z = 7
# x + 3y + 2z = 13
# x + 5z = 16

A = np.array([[2., 1., 1.],
              [1., 3., 2.],
              [1., 0., 5.]])
b = np.array([7., 13., 16.])

M = np.hstack([A, b.reshape(-1, 1)])
n = 3

print("Matriks Awal:")
print(M)

for k in range(n):
    pivot = np.argmax(np.abs(M[k:, k])) + k
    M[[k, pivot]] = M[[pivot, k]]
    for i in range(k+1, n):
        m = M[i, k] / M[k, k]
        M[i, :] = M[i, :] - m * M[k, :]

print("\nMatriks Segitiga Atas:")
print(M)

x = np.zeros(n)
for i in range(n-1, -1, -1):
    x[i] = (M[i, n] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]

print("\n=== HASIL AKHIR ===")
print(f"x = {x[0]:.0f}")
print(f"y = {x[1]:.0f}")
print(f"z = {x[2]:.0f}")
print(f"\nCek: 2(1)+2+3={2*x[0]+x[1]+x[2]:.0f}, 1+6+6={x[0]+3*x[1]+2*x[2]:.0f}, 1+15={x[0]+5*x[2]:.0f}")
```
